```python
import pandas as pd
from sklearn import tree
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../Data/barbecue.csv")
data.head()
```

```python
data['barbecue']
```

```python
X_train = data.drop('barbecue', axis=1)
y_train = data['barbecue']
```

```python
model = tree.DecisionTreeClassifier(criterion='entropy')
history = model.fit(X_train, y_train)
```

```python
feature_names=['Meteo', 'Amis', 'Vent', 'Jour']
target_names=['non', 'oui']
```

```python
plt.figure(figsize=(20, 15))
plot_tree(
    model,
    feature_names=feature_names,
    class_names=target_names,
    filled=True, # couleurs
    fontsize=17
)
plt.title("Arbre de décision barbecue", fontsize=20)
plt.savefig('decision_tree_barbecue.png')
```
